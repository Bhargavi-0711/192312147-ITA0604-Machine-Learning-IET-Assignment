```python
# ============================================================
# MACHINE LEARNING-BASED WATER QUALITY SCREENING
# Candidate Elimination + Decision Tree Learning
# ============================================================

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import LabelEncoder
from sklearn.tree import DecisionTreeClassifier, plot_tree, export_text
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    confusion_matrix,
    classification_report
)

# ============================================================
# 1. CREATE EDUCATIONAL WATER-QUALITY DATASET
# ============================================================

data = {
    "pH": [
        "Neutral", "Neutral", "Acidic", "Neutral", "Alkaline",
        "Acidic", "Neutral", "Alkaline", "Neutral", "Acidic",
        "Neutral", "Alkaline", "Neutral", "Acidic", "Neutral",
        "Alkaline", "Neutral", "Acidic", "Neutral", "Alkaline",
        "Neutral", "Neutral", "Acidic", "Alkaline", "Neutral",
        "Neutral", "Acidic", "Neutral", "Alkaline", "Neutral"
    ],

    "Turbidity": [
        "Low", "Low", "High", "Low", "Low",
        "High", "Low", "High", "Low", "High",
        "Low", "Low", "Low", "High", "Low",
        "High", "Low", "High", "Low", "High",
        "Low", "Low", "High", "Low", "Low",
        "High", "High", "Low", "High", "Low"
    ],

    "TDS": [
        "Normal", "Normal", "High", "Normal", "Normal",
        "High", "Normal", "High", "Normal", "High",
        "Normal", "Normal", "Normal", "High", "Normal",
        "High", "Normal", "High", "Normal", "High",
        "Normal", "Normal", "High", "Normal", "Normal",
        "High", "High", "Normal", "High", "Normal"
    ],

    "Chlorine": [
        "Adequate", "Adequate", "Low", "Adequate", "Adequate",
        "Low", "Adequate", "Low", "Adequate", "Low",
        "Adequate", "Adequate", "Adequate", "Low", "Adequate",
        "Low", "Adequate", "Low", "Adequate", "Low",
        "Adequate", "Adequate", "Low", "Adequate", "Adequate",
        "Low", "Low", "Adequate", "Low", "Adequate"
    ],

    "Microbial": [
        "Absent", "Absent", "Present", "Absent", "Absent",
        "Present", "Absent", "Present", "Absent", "Present",
        "Absent", "Absent", "Absent", "Present", "Absent",
        "Present", "Absent", "Present", "Absent", "Present",
        "Absent", "Absent", "Present", "Absent", "Absent",
        "Present", "Present", "Absent", "Present", "Absent"
    ],

    "Source": [
        "Borewell", "Municipal", "River", "Municipal", "Borewell",
        "River", "Municipal", "River", "Borewell", "River",
        "Municipal", "Borewell", "Municipal", "River", "Borewell",
        "River", "Municipal", "River", "Borewell", "River",
        "Municipal", "Borewell", "River", "Municipal", "Borewell",
        "River", "River", "Municipal", "River", "Borewell"
    ],

    "Class": [
        "Safe", "Safe", "Needs Treatment", "Safe", "Safe",
        "Needs Treatment", "Safe", "Needs Treatment", "Safe", "Needs Treatment",
        "Safe", "Safe", "Safe", "Needs Treatment", "Safe",
        "Needs Treatment", "Safe", "Needs Treatment", "Safe", "Needs Treatment",
        "Safe", "Safe", "Needs Treatment", "Safe", "Safe",
        "Needs Treatment", "Needs Treatment", "Safe", "Needs Treatment", "Safe"
    ]
}

df = pd.DataFrame(data)

print("=" * 60)
print("WATER QUALITY DATASET")
print("=" * 60)
print(df)

print("\nDataset Shape:", df.shape)
print("\nClass Distribution:")
print(df["Class"].value_counts())


# ============================================================
# 2. PREPROCESSING
# ============================================================

features = [
    "pH",
    "Turbidity",
    "TDS",
    "Chlorine",
    "Microbial",
    "Source"
]

target = "Class"

# Encode categorical features
encoders = {}

df_encoded = df.copy()

for column in features:
    encoder = LabelEncoder()
    df_encoded[column] = encoder.fit_transform(df_encoded[column])
    encoders[column] = encoder

# Encode target
target_encoder = LabelEncoder()
df_encoded[target] = target_encoder.fit_transform(df_encoded[target])

print("\nEncoded Dataset:")
print(df_encoded.head())


# ============================================================
# 3. TRAIN-TEST SPLIT
# ============================================================

X = df_encoded[features]
y = df_encoded[target]

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.30,
    random_state=42,
    stratify=y
)

print("\nTraining Samples:", len(X_train))
print("Testing Samples :", len(X_test))


# ============================================================
# 4. DECISION TREE MODEL
# ============================================================

tree_model = DecisionTreeClassifier(
    criterion="entropy",
    max_depth=4,
    random_state=42
)

tree_model.fit(X_train, y_train)

y_pred = tree_model.predict(X_test)


# ============================================================
# 5. EVALUATION
# ============================================================

accuracy = accuracy_score(y_test, y_pred)

precision = precision_score(
    y_test,
    y_pred,
    average="weighted",
    zero_division=0
)

recall = recall_score(
    y_test,
    y_pred,
    average="weighted",
    zero_division=0
)

f1 = f1_score(
    y_test,
    y_pred,
    average="weighted",
    zero_division=0
)

print("\n" + "=" * 60)
print("DECISION TREE EVALUATION")
print("=" * 60)

print("Accuracy  :", round(accuracy, 4))
print("Precision :", round(precision, 4))
print("Recall    :", round(recall, 4))
print("F1 Score  :", round(f1, 4))

print("\nClassification Report:")
print(
    classification_report(
        y_test,
        y_pred,
        target_names=target_encoder.classes_,
        zero_division=0
    )
)


# ============================================================
# 6. CONFUSION MATRIX
# ============================================================

cm = confusion_matrix(y_test, y_pred)

print("\nConfusion Matrix:")
print(cm)

plt.figure(figsize=(6, 5))
plt.imshow(cm)
plt.title("Confusion Matrix - Water Quality Screening")
plt.xlabel("Predicted Class")
plt.ylabel("Actual Class")
plt.xticks(
    range(len(target_encoder.classes_)),
    target_encoder.classes_
)
plt.yticks(
    range(len(target_encoder.classes_)),
    target_encoder.classes_
)

for i in range(cm.shape[0]):
    for j in range(cm.shape[1]):
        plt.text(j, i, cm[i, j],
                 ha="center",
                 va="center")

plt.colorbar()
plt.show()


# ============================================================
# 7. DISPLAY DECISION TREE
# ============================================================

plt.figure(figsize=(18, 10))

plot_tree(
    tree_model,
    feature_names=features,
    class_names=target_encoder.classes_,
    filled=True,
    rounded=True
)

plt.title("Decision Tree for Water Quality Screening")
plt.show()


# ============================================================
# 8. PRINT DECISION TREE RULES
# ============================================================

print("\n" + "=" * 60)
print("DECISION TREE RULES")
print("=" * 60)

rules = export_text(
    tree_model,
    feature_names=features
)

print(rules)


# ============================================================
# 9. MANUAL ENTROPY FUNCTION
# ============================================================

def entropy(labels):

    values, counts = np.unique(labels, return_counts=True)

    probabilities = counts / len(labels)

    entropy_value = 0

    for p in probabilities:
        if p > 0:
            entropy_value -= p * np.log2(p)

    return entropy_value


# ============================================================
# 10. MANUAL INFORMATION GAIN
# ============================================================

def information_gain(data, feature, target):

    total_entropy = entropy(data[target])

    weighted_entropy = 0

    for value in data[feature].unique():

        subset = data[data[feature] == value]

        weight = len(subset) / len(data)

        weighted_entropy += weight * entropy(subset[target])

    gain = total_entropy - weighted_entropy

    return gain


print("\n" + "=" * 60)
print("MANUAL INFORMATION GAIN CALCULATION")
print("=" * 60)

dataset_entropy = entropy(df[target])

print("Dataset Entropy =", round(dataset_entropy, 4))

information_gains = {}

for feature in features:

    gain = information_gain(
        df,
        feature,
        target
    )

    information_gains[feature] = gain

    print(
        feature,
        "Information Gain =",
        round(gain, 4)
    )

best_feature = max(
    information_gains,
    key=information_gains.get
)

print("\nBest Root Feature:")
print(best_feature)


# ============================================================
# 11. CANDIDATE ELIMINATION
# ============================================================

print("\n" + "=" * 60)
print("CANDIDATE ELIMINATION")
print("=" * 60)

# Use a small training subset for concept learning
ce_data = df.iloc[:12].copy()

attribute_values = {}

for feature in features:
    attribute_values[feature] = list(
        df[feature].unique()
    )

# Initial Specific hypothesis
S = ["Ø"] * len(features)

# Initial General hypothesis
G = [["?"] * len(features)]

print("\nInitial Specific Boundary S:")
print(S)

print("\nInitial General Boundary G:")
print(G)


# ============================================================
# Candidate Elimination Functions
# ============================================================

def covers(hypothesis, example):

    for h, value in zip(hypothesis, example):

        if h == "?":
            continue

        if h == "Ø":
            return False

        if h != value:
            return False

    return True


def more_general_or_equal(h1, h2):

    for a, b in zip(h1, h2):

        if a == "?":
            continue

        if a == b:
            continue

        return False

    return True


def generalize_S(S, example):

    new_S = S.copy()

    for i in range(len(S)):

        if S[i] == "Ø":
            new_S[i] = example[i]

        elif S[i] != example[i]:
            new_S[i] = "?"

    return new_S


def specialize_G(G, example, attribute_values):

    new_G = []

    for hypothesis in G:

        for i in range(len(hypothesis)):

            if hypothesis[i] == "?":

                for value in attribute_values[features[i]]:

                    if value != example[i]:

                        new_h = hypothesis.copy()

                        new_h[i] = value

                        new_G.append(new_h)

    return new_G


# ============================================================
# RUN CANDIDATE ELIMINATION
# ============================================================

for index, row in ce_data.iterrows():

    example = [
        row[feature]
        for feature in features
    ]

    label = row[target]

    print("\n----------------------------------------")
    print("Example", index + 1)
    print("Features:", example)
    print("Class:", label)

    if label == "Safe":

        # Generalize S
        S = generalize_S(S, example)

        # Remove G hypotheses that do not cover example
        G = [
            g for g in G
            if covers(g, example)
        ]

    else:

        # Specialize G
        G = specialize_G(
            G,
            example,
            attribute_values
        )

        # Remove hypotheses that cover negative example
        G = [
            g for g in G
            if not covers(g, example)
        ]

    print("S =", S)

    print("G =")

    for g in G[:10]:
        print(g)

    if len(G) > 10:
        print("... more hypotheses ...")


# ============================================================
# 12. FINAL VERSION SPACE
# ============================================================

print("\n" + "=" * 60)
print("FINAL VERSION SPACE")
print("=" * 60)

print("\nSpecific Boundary S:")
print(S)

print("\nGeneral Boundary G:")

if len(G) == 0:
    print("No consistent general hypotheses found.")
else:
    for g in G[:20]:
        print(g)

print("\nNumber of General Hypotheses:", len(G))


# ============================================================
# 13. SAMPLE WATER PREDICTION
# ============================================================

print("\n" + "=" * 60)
print("NEW WATER SAMPLE PREDICTION")
print("=" * 60)

new_sample = pd.DataFrame({
    "pH": ["Neutral"],
    "Turbidity": ["Low"],
    "TDS": ["Normal"],
    "Chlorine": ["Adequate"],
    "Microbial": ["Absent"],
    "Source": ["Borewell"]
})

print("\nNew Water Sample:")
print(new_sample)

# Encode using previously trained encoders
new_encoded = new_sample.copy()

for feature in features:
    new_encoded[feature] = encoders[feature].transform(
        new_encoded[feature]
    )

prediction = tree_model.predict(
    new_encoded[features]
)

predicted_class = target_encoder.inverse_transform(
    prediction
)

print("\nPredicted Result:")
print(predicted_class[0])


# ============================================================
# 14. FINAL CONCLUSION
# ============================================================

print("\n" + "=" * 60)
print("CONCLUSION")
print("=" * 60)

print("""
The developed educational machine-learning system classifies
water samples into:

1. Safe
2. Needs Treatment

Candidate Elimination demonstrates how hypotheses change when
positive and negative examples are observed.

The Decision Tree provides an interpretable set of IF-THEN
rules using entropy and information gain.

This model is intended only for educational screening and
must NOT replace certified laboratory water-quality testing.
""")
```
